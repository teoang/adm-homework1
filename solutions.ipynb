## 6. Write a function

````python
def is_leap(year):
    leap = False
    
    # Write your logic here
    
    if((year%4)==0):
        if((year%100)==0 and (year%400)==0):
            leap=True
            return leap
        if((year%100)==0):
            leap=False
            return leap
        leap=True
    
    return leap

year = int(input())
print(is_leap(year))
````

## 8. List Comprehensions

[Accepted submission](https://www.hackerrank.com/challenges/list-comprehensions/submissions/code/484655655) - PyPy 3

**Example input**

```text
1
1
1
2
```

````python
if __name__ == '__main__':
    x = int(input())
    y = int(input())
    z = int(input())
    n = int(input())
    
    lista = [[i,j,k] for i in range(0,x+1) for j in range(0,y+1) for k in range(0,z+1) if i+j+k!=n]

    print(lista)
````

## 9. Find the Runner-Up Score!

[Accepted submission](https://www.hackerrank.com/challenges/find-second-maximum-number-in-a-list/submissions/code/484980362) - PyPy 3

**Example input**

```text
5
2 3 6 6 5
```

````python
if __name__ == '__main__':
    n = int(input())
    arr = map(int, input().split())

    first = second = float('-inf')

    for x in arr:
        if x > first:
            second = first
            first = x
        elif first > x > second:
            second = x

    print(second)
````
